```python
df['loan_approved'].value_counts().plot(kind='bar')

plt.title("Loan Approval Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.hist(df['credit_score'], bins=10)

plt.xlabel("Credit Score")
plt.ylabel("Frequency")
plt.title("Credit Score Distribution")
plt.show()
```

```python
sns.scatterplot(x='income',
    y='loan_amount',
    hue='loan_approved',
    data=df
)

plt.title('Income vs Loan Amount')
plt.show()
```

```python
sns.boxplot(
    x='loan_approved',
    y='credit_score',
    data=df
)

plt.title('Credit Score by Loan Approval')
plt.show()
```

```python
sns.barplot(
    x='years_employed',
    y='loan_approved',
    data=df
)

plt.xticks(rotation=45)
plt.title('Average Loan Amount by City')
plt.xticks(rotation=90)
plt.show()
```

```python
counts=df["loan_approved"].value_counts()
plt.pie(counts,labels=counts.index,autopct="%1.1f%%")
plt.title("approval of loan")
plt.show()

```

```python
loan_status=df.groupby("loan_approved")[[
           "loan_amount",
           "income",
           "credit_score"
]].mean()
loan_status.plot(kind="bar")
plt.title("loan approval analysis")
plt.xlabel("loan status")
plt.ylabel("average value") 
plt.show()          
```

```python
sns.relplot(x="income",y="loan_amount",hue="loan_approved",data=df)
plt.show()
```

```python
sns.catplot(x="loan_approved",y="credit_score",kind="bar",data=df)
plt.show()
```
